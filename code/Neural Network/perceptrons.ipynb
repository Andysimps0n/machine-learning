```python
from sklearn.datasets import load_iris
import numpy as np
from sklearn.linear_model import Perceptron
```

```python
data = load_iris()

X = data.data[:, (2,3)]
y = (data.target == 0 ).astype(np.int)      
```

```python
per_clf = Perceptron()
per_clf.fit(X,y)
```

```python
y_pred = per_clf.predict([[2, 0.5]])
y_pred
```
