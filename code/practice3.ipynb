```python
import pandas as pd
import numpy as np  
import matplotlib.pyplot as plt     
from sklearn.model_selection import train_test_split, StratifiedShuffleSplit

data = pd.read_csv('../data/housing.csv')

# median_income => median_house_value
```

```python
# Splitting the data

data['median_income_cat'] = pd.qcut(data["median_income"], q=5, labels=False)
data_splitter = StratifiedShuffleSplit(n_splits=1, test_size=0.2, random_state=3)

for train_index, test_index in data_splitter.split(data, data["median_income_cat"]):
    strat_train_set = data.loc[train_index]
    strat_test_set = data.loc[test_index]

```

```python
# Detatching the target value
housing = strat_train_set.drop("median_house_value", axis=1)
housing_labes = strat_train_set["median_house_value"].copy()
```

```python
# Pipeline : imputer, scaler

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer


num_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy="median")),
    ('std_scaler', StandardScaler())
])

housing_num_tr = num_pipeline.fit_transform(housing)
```
