```python
from sklearn.datasets import fetch_openml
mnist = fetch_openml('mnist_784', as_frame=False)
mnist.keys()
```

```python
from sklearn.model_selection import train_test_split
import numpy as np
X, y = mnist["data"], mnist["target"]

y = y.astype(np.uint8)
y = (y==5)


# Train and test
X_train, X_test_valid, y_train, y_test_valid = train_test_split(X, y, test_size=0.3, random_state=39)


# Validation and test
X_valid, X_test, y_valid, y_test = train_test_split(X_test_valid, y_test_valid, test_size=0.5, random_state=39)
```
