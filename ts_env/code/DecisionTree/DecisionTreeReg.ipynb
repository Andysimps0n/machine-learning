```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeRegressor

data = load_iris()
X = data.data[:, :2]
y = data.data[:, 3]
```

```python
model = DecisionTreeRegressor(    
    max_depth=None,        # no depth limit
    min_samples_split=2,   # default, split until pure
    min_samples_leaf=1,    # default, smallest possible leaf
    max_features=None,     # use all features
    random_state=42
    )
model.fit(X, y)
```

```python
from sklearn.tree import DecisionTreeRegressor
import numpy as np
import matplotlib.pyplot as plt

X1 = X[:, [0]]  # use just one feature
reg = DecisionTreeRegressor(max_depth=2).fit(X1, y)

# sort X for plotting
x_range = np.linspace(X1.min(), X1.max(), 500).reshape(-1, 1)
y_pred = reg.predict(x_range)

plt.scatter(X1, y, c="blue", alpha=0.5)
plt.plot(x_range, y_pred, color="red", linewidth=2)
plt.xlabel(data.feature_names[0])
plt.ylabel("Target")
plt.title("Decision Tree Regressor (stepwise fit)")
plt.show()
```
