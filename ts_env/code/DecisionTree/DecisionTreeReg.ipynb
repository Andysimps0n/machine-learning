```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeRegressor

data = load_iris()
X = data.data[:, :2]
y = data.target
```

```python
model = DecisionTreeRegressor(    
    max_depth=2,        # no depth limit
    random_state=42
    )
model.fit(X, y)
```

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.tree import DecisionTreeRegressor


x_min, x_max = X[:, 0].min() - 0.5, X[:, 0].max() + 0.5
y_min, y_max = X[:, 1].min() - 0.5, X[:, 1].max() + 0.5
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 200),
                     np.linspace(y_min, y_max, 200))

Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

# Plot
plt.figure(figsize=(8,6))
plt.scatter(X[:, 0], X[:, 1], c=y, edgecolor="k", cmap=plt.cm.RdBu)
plt.xlabel(data.feature_names[0])
plt.contourf(xx, yy, Z, alpha=0.3, cmap=plt.cm.RdBu)

plt.ylabel(data.feature_names[1])
plt.title("Decision Tree Regression on Iris data")
plt.colorbar(label="Predicted petal width")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeRegressor

# Use only sepal length (first column)
data = load_iris()
X = data.data[:, 0].reshape(-1, 1)  # sepal length
y = data.data[:, 3]                 # petal width

# Train model
model = DecisionTreeRegressor(max_depth=3, random_state=42)
model.fit(X, y)

# Test data (for smooth line)
X_test = np.linspace(X.min(), X.max(), 200).reshape(-1, 1)
y_pred = model.predict(X_test)

# Plot
plt.scatter(X, y, s=20, edgecolor="black", c="orange", label="data")
plt.plot(X_test, y_pred, color="green", linewidth=2, label="prediction")
plt.xlabel("Sepal length")
plt.ylabel("Petal width")
plt.title("Decision Tree Regression (Iris)")
plt.legend()
plt.show()
```
