```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import mean_squared_error
import pandas as pd


# without scaling MSE : 915
```

```python
train_full = pd.read_csv('./train.csv') 
y_full = pd.read_csv('./test.csv') 


train_full.drop(['Id'], axis=1, inplace=True)
train_full = train_full.select_dtypes(include=['float64', 'int64'])
X = train_full.drop(columns=['SalePrice'])
y = train_full['SalePrice']
X_train, X_valid, y_train, y_valid = train_test_split(X, y, test_size=0.3, random_state=42)

X_train
```

```python
y_full
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GridSearchCV

model = RandomForestRegressor(random_state=42)

param_grid = {
    "n_estimators": [100, 300],
    "max_depth": [3, 5],
    "min_samples_split": [2, 4],
    "min_samples_leaf": [1, 2]
}

grid_search = GridSearchCV(
    estimator=model,
    param_grid=param_grid,
    cv=3,
    n_jobs=-1
)

grid_search.fit(X_train, y_train)
```

```python
model = grid_search.best_estimator_

y_pred = grid_search.predict(X_valid)
mse = mean_squared_error(y_valid, y_pred)
print("Validation MSE:", mse/1000000)
```

```python
test_data = pd.read_csv("./test.csv")
test_data = test_data[X_train.columns]  # keep only the numeric columns the model was trained on

submission_data = grid_search.predict(test_data)

submission_df = pd.DataFrame({
    'Id': y_full['Id'],
    'SalePrice': submission_data
})

submission_df.to_csv('submission_5.csv', index=False)
submission_df
```
